```python
import pandas as pd
```

```python
data= pd.read_csv('/content/predictive_maintenance_atms.csv')
data.head()
```

```python
def impute_missing_values(df):
    # Handle missing values
    categorical_columns = ['Location_Type', 'Manufacturer', 'Service_Contract_Type']
    numerical_columns = ['Year_of_Installation', 'Frequency_of_Maintenance', 'Downtime_Duration_hours',
                         'Number_of_Incidents', 'Maintenance_Costs_USD', 'Transaction_Volumes',
                         'Average_Withdrawal_Amount', 'Card_Reader_Error_Rate_percent',
                         'Cash_Dispenser_Jam_Rate_percent', 'ATM_Age_Years']

    # Add missing categorical values with mode
    for column in categorical_columns:
        mode_value = df[column].mode().iloc[0]
        df[column].fillna(mode_value, inplace=True)

    # Add missing numerical values with median
    for column in numerical_columns:
        median_value = df[column].median()
        df[column].fillna(median_value, inplace=True)

    # Add missing date values with the most frequent date
    if 'Last_Maintenance_Date' in df.columns:
        most_frequent_date = df['Last_Maintenance_Date'].mode().iloc[0]
        df['Last_Maintenance_Date'].fillna(most_frequent_date, inplace=True)

    # Fill missing ATM_ID with a unique identifier based on the row index
    if 'ATM_ID' in df.columns:
        df['ATM_ID'] = df['ATM_ID'].fillna(df.index.to_series().apply(lambda x: f'ATM_{x+1}'))

    return df

# Applying the changes to the dataset
data = impute_missing_values(data)

# Verifying
missing_values_after_imputation = data.isnull().sum()
missing_values_after_imputation
```

```python
numerical_columns = ['Year_of_Installation', 'Frequency_of_Maintenance', 'Downtime_Duration_hours',
                         'Number_of_Incidents', 'Maintenance_Costs_USD', 'Transaction_Volumes',
                         'Average_Withdrawal_Amount', 'Card_Reader_Error_Rate_percent',
                         'Cash_Dispenser_Jam_Rate_percent', 'ATM_Age_Years']
```

```python
data[numerical_columns] = data[numerical_columns].round(2)
```

```python
#Save the file in CSV
output_file_path = 'modified_predictive_maintenance_atms.csv'
data.to_csv(output_file_path, index=False)
```
